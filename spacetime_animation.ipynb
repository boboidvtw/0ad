```python
%%manim -v WARNING -qm SpacetimeMassAnimation

from manim import *
import numpy as np

class SpacetimeMassAnimation(ThreeDScene):
    def construct(self):

        # -----------------------------------------
        # Configuration and Setup
        # -----------------------------------------

        COLOR_T1 = TEAL        # Macro Cosmic Time
        COLOR_T2 = PURPLE_C    # Micro Quantum Phase Time
        COLOR_T3 = YELLOW      # Measurement Axis
        COLOR_SPACE = WHITE    # Spatial Axes

        # 標題
        title = Tex(r"(3+3) Spacetime \& Rest Mass Origin", font_size=40)
        title.to_edge(UP)

        # -----------------------------------------
        # Step 1: Axes Setup
        # -----------------------------------------

        axes = ThreeDAxes(
            x_range=[-4, 4, 1],
            y_range=[-4, 4, 1],
            z_range=[-4, 4, 1],
            x_length=6, y_length=6, z_length=6,
            axis_config={
                "include_tip": True,
                "include_ticks": True,
                "stroke_width": 2,
            },
        )

        # 空間軸標籤
        x_label = axes.get_x_axis_label(Tex("X", color=COLOR_SPACE))
        y_label = axes.get_y_axis_label(Tex("Y", color=COLOR_SPACE))
        z_label = axes.get_z_axis_label(Tex("Z", color=COLOR_SPACE))
        spatial_labels = VGroup(x_label, y_label, z_label)

        # 時間軸
        t1_axis = Line(start=axes.c2p(-4, 0, 0), end=axes.c2p(4, 0, 0), color=COLOR_T1, stroke_width=4)
        t2_axis = Line(start=axes.c2p(0, -4, 0), end=axes.c2p(0, 4, 0), color=COLOR_T2, stroke_width=4)
        t3_axis = Line(start=axes.c2p(0, 0, -4), end=axes.c2p(0, 0, 4), color=COLOR_T3, stroke_width=4)

        # 軸箭頭
        t1_tip = Cone(direction=RIGHT, resolution=8).scale(0.08).set_color(COLOR_T1).move_to(axes.c2p(4, 0, 0))
        t2_tip = Cone(direction=UP, resolution=8).scale(0.08).set_color(COLOR_T2).move_to(axes.c2p(0, 4, 0))
        t3_tip = Cone(direction=OUT, resolution=8).scale(0.08).set_color(COLOR_T3).move_to(axes.c2p(0, 0, 4))

        t_axes_g = VGroup(t1_axis, t2_axis, t3_axis, t1_tip, t2_tip, t3_tip)

        # 時間軸標籤
        t1_label = MathTex(r"t_1", color=COLOR_T1).next_to(t1_tip, RIGHT)
        t2_label = MathTex(r"t_2", color=COLOR_T2).next_to(t2_tip, UP)
        t3_label = MathTex(r"t_3", color=COLOR_T3).next_to(t3_tip, OUT)
        temporal_labels = VGroup(t1_label, t2_label, t3_label)

        # 鏡頭角度設定
        self.set_camera_orientation(phi=75 * DEGREES, theta=-45 * DEGREES)

        # 動畫開場
        self.add_fixed_in_frame_mobjects(title)
        self.play(Write(title))

        self.play(
            Create(axes),
            Write(spatial_labels),
            run_time=2
        )
        self.play(
            GrowFromCenter(t_axes_g),
            Write(temporal_labels),
            run_time=2
        )
        self.wait(1)

        # -----------------------------------------
        # Step 2: System Photon Trapping
        # -----------------------------------------

        trapping_point_coords = np.array([2.0, 2.0, 2.0])
        point_3d = axes.c2p(*trapping_point_coords)

        dot = Dot(point_3d, color=YELLOW, radius=0.1)
        target_cross = VGroup(
            Line(point_3d + LEFT*0.3, point_3d + RIGHT*0.3, color=YELLOW, stroke_width=2),
            Line(point_3d + DOWN*0.3, point_3d + UP*0.3, color=YELLOW, stroke_width=2),
            Line(point_3d + OUT*0.3, point_3d + IN*0.3, color=YELLOW, stroke_width=2)
        )

        self.play(FadeIn(dot), Create(target_cross), run_time=1.5)

        photon = Dot(point_3d, color=RED, radius=0.15)
        photon_label = MathTex(r"\gamma_{sys}", color=RED, font_size=36).next_to(photon, UP)
        self.play(FadeIn(photon), Write(photon_label))

        # -----------------------------------------
        # Step 3: t2 Rotation & Oscillation
        # -----------------------------------------

        self.play(axes.animate.shift(LEFT*2 + IN*2), run_time=2)
        self.move_camera(phi=60 * DEGREES, theta=-30 * DEGREES, run_time=2)

        point_3d_new = axes.c2p(*trapping_point_coords)
        self.play(
            photon.animate.move_to(point_3d_new),
            photon_label.animate.next_to(photon, UP),
            target_cross.animate.move_to(point_3d_new),
            run_time=0.5
        )

        osc_radius = 0.8
        circle_center = axes.c2p(0, trapping_point_coords[1], 0)
        orbit_circle = Circle(radius=osc_radius, color=COLOR_T2, stroke_opacity=0.6).move_to(circle_center)

        orbit_start_point = axes.c2p(osc_radius, trapping_point_coords[1], 0)
        self.play(photon.animate.move_to(orbit_start_point), run_time=1)

        self.add(orbit_circle)
        # 修正：將 radians 改為 angle
        self.play(
            Rotating(
                photon,
                angle=2*PI * 3,
                axis=UP,
                about_point=circle_center
            ),
            run_time=3,
            rate_func=linear
        )

        # -----------------------------------------
        # Step 4: Emergence of Mass
        # -----------------------------------------

        current_photon_pos = photon.get_center()

        mass_eq = MathTex(r"E_2 = m_0 c^2", color=WHITE, font_size=48)
        mass_eq.to_corner(DR)
        self.add_fixed_in_frame_mobjects(mass_eq)
        self.play(Write(mass_eq), run_time=1.5)

        arrow_start = circle_center + LEFT*0.2
        arrow_end = current_photon_pos + RIGHT*0.1 + UP*0.1
        mass_arrow = Arrow(arrow_start, arrow_end, color=COLOR_T2, buff=0.1)
        mass_label = MathTex(r"\text{Rest Mass } (m_0)", color=COLOR_T2, font_size=30).next_to(mass_arrow, LEFT)

        self.play(Create(mass_arrow), Write(mass_label))
        self.wait(2)

        # -----------------------------------------
        # Step 5: Final Camera Rotation
        # -----------------------------------------

        scene_elements = VGroup(
            axes, t_axes_g, spatial_labels, temporal_labels,
            dot, target_cross, photon, photon_label,
            orbit_circle, mass_arrow, mass_label
        )

        self.begin_ambient_camera_rotation(rate=0.2 * PI)

        self.play(
            FadeOut(mass_eq),
            FadeOut(scene_elements),
            run_time=4
        )

        self.wait(2)
        self.stop_ambient_camera_rotation()
```
